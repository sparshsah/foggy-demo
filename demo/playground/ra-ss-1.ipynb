```python
"""
Prompt:

Convert a list of lists into a dictionary tree.

Given a list of lists, convert it into a dictionary of dictionaries,
with the final layer of the dictionary being a tuple.

Examples:

labels = [
["A", "I"],
["A", "II"],
["B", "I"],
["B", "II"],
]

tree = to_tree(labels)
assert tree == {
"A": ("I", "II"),
"B": ("I", "II")
}

labels = [
("A", "I", 'a'),
("A", "I", 'b'),
("A", "I", 'c'),

("A", "II", 'a'),

("A", "III", 'a'),
("A", "III", 'b'),
("A", "III", 'c'),
("A", "III", 'd'),

("B", "I", 'a'),
("B", "I", 'b'),

("B", "II", 'a'),

("B", "III", 'a'),
("B", "III", 'b'),
("B", "III", 'c'),
]

tree = to_tree(labels)
assert tree == {
'A': {
'I': ('a', 'b', 'c'),
'II': ('a',),
'III': ('a', 'b', 'c', 'd'),
},
'B': {
'I': ('a', 'b'),
'II': ('a',),
'III': ('a', 'b', 'c'),
},
}
"""
```

```python
def _to_tree_2d(labels: list[tuple[str, str]]) -> dict[str, tuple]:
    """
    Convert a list of lists into a dict.
    Helper for 2d case.
    """
    tree = {}
    for label in labels:
        key, value = label
        if key not in tree:
            # initialize key with singleton value
            tree[key] = (value,)
        else:
            # append value
            tree[key] += (value,)
    return tree


def _to_tree_3d(labels: list[tuple[str, str, str]]) -> dict[str, dict[str, tuple]]:
    """
    Helper for 3d case.
    """
    tree = {}
    for label in labels:
        outer_key, inner_key, value = label
        if outer_key not in tree:
            # initialize inner dict
            tree[outer_key] = {}
            tree[outer_key][inner_key] = (value,)
        elif outer_key in tree and inner_key not in tree[outer_key]:
            assert isinstance(tree[outer_key], dict), f"Malformed type! {type(tree[outer_key])}"
            tree[outer_key][inner_key] = (value,)
        else:
            # both the outer_key and inner_key are in the tree,
            # so preserve the existing values and append the new one
            tree[outer_key][inner_key] += (value,)
    return tree


def to_tree(labels: list) -> dict:
    # TODO(sparsh): Try a recursive implementation.
    # TODO(sparsh): Use DefaultDict with empty tuple to reduce branches.
    # TODO(sparsh): Keep track as linked list instead of reallocating new tuple each iteration.
    tree = {}
    for label in labels:
        if len(label) < 2:
            raise ValueError(f"Malformed label! Expected len-2-or-more, got {label}.")
        *outer_keys, inner_key, value = label
        # keep track of the current level of the tree
        curr_subtree = tree
        for outer_key in outer_keys:
            if outer_key not in curr_subtree:
                curr_subtree[outer_key] = {}
            curr_subtree = curr_subtree[outer_key]
        # handle the inner key and value
        if inner_key not in curr_subtree:
            curr_subtree[inner_key] = (value,)
        else:
            curr_subtree[inner_key] += (value,)
    return tree
```

```python
labels = [
    ["A", "I"],
    ["A", "II"],
    ["B", "I"],
    ["B", "II"],
]
print(_to_tree_2d(labels))
print(to_tree(labels))
```

```python
labels = [
    ("A", "I", 'a'),
    ("A", "I", 'b'),
    ("A", "I", 'c'),

    ("A", "II", 'a'),

    ("A", "III", 'a'),
    ("A", "III", 'b'),
    ("A", "III", 'c'),
    ("A", "III", 'd'),

    ("B", "I", 'a'),
    ("B", "I", 'b'),

    ("B", "II", 'a'),

    ("B", "III", 'a'),
    ("B", "III", 'b'),
    ("B", "III", 'c'),
]
print(_to_tree_3d(labels))
print(to_tree(labels))
```
